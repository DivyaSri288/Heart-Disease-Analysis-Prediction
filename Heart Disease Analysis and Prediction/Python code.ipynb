## 6. Data Cleaning and Preprocessing

In this step, the dataset is cleaned and prepared for further
analysis and machine learning.

The preprocessing steps include:
- Checking missing values
- Checking duplicate records
- Handling missing values
- Separating features and target variable
- Preparing the data for machine learning

```python
# Check missing values

df.isnull().sum()
```

```python
# Check duplicate records

print("Duplicate records before cleaning:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate records after cleaning:", df.duplicated().sum())
```

```python
# Handle missing values

for column in df.columns:
    if df[column].isnull().sum() > 0:
        if df[column].dtype == 'object':
            df[column] = df[column].fillna(df[column].mode()[0])
        else:
            df[column] = df[column].fillna(df[column].median())

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Final dataset shape:", df.shape)

df.head()
```

## 10. Data Preprocessing for Machine Learning

Before training the machine learning models, the input features are
converted into a suitable numerical format.

Categorical variables are encoded, numerical variables are scaled,
and the dataset is divided into training and testing sets.

```python
# Identify categorical and numerical columns

categorical_columns = X.select_dtypes(include=['object', 'category']).columns
numerical_columns = X.select_dtypes(include=['int64', 'float64']).columns

print("Categorical columns:")
print(list(categorical_columns))

print("\nNumerical columns:")
print(list(numerical_columns))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_columns),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_columns)
    ]
)

X_processed = preprocessor.fit_transform(X)

print("Original feature shape:", X.shape)
print("Processed feature shape:", X_processed.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_processed,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

### Train-Test Split

The dataset is divided into 80% training data and 20% testing data.
The training data is used to build the machine learning models, while
the testing data is used to evaluate their performance on unseen data.

## 11. Machine Learning Model Building

In this section, machine learning algorithms are trained to predict
the presence of heart disease.

The first model used is Logistic Regression, which is a commonly used
classification algorithm for binary prediction problems.

```python
from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression model
logistic_model = LogisticRegression(max_iter=1000, random_state=42)

# Train the model
logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Make predictions on test data

y_pred_logistic = logistic_model.predict(X_test)

print("Predictions generated successfully!")
print(y_pred_logistic[:10])
```

### Logistic Regression

The Logistic Regression model is trained using the training dataset
and then used to predict heart disease outcomes for the unseen
testing dataset.

## 12. Decision Tree and Random Forest

To compare different classification approaches, Decision Tree and
Random Forest algorithms are also trained on the processed dataset.

Decision Tree uses a tree-like structure to make predictions, while
Random Forest combines multiple decision trees to improve prediction
performance.

```python
from sklearn.tree import DecisionTreeClassifier

# Create Decision Tree model
decision_tree_model = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

# Train the model
decision_tree_model.fit(X_train, y_train)

# Make predictions
y_pred_tree = decision_tree_model.predict(X_test)

print("Decision Tree model trained successfully!")
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
random_forest_model.fit(X_train, y_train)

# Make predictions
y_pred_forest = random_forest_model.predict(X_test)

print("Random Forest model trained successfully!")
```

### Model Training Summary

Three classification models have now been trained:

1. Logistic Regression
2. Decision Tree
3. Random Forest

The models will be evaluated using accuracy, precision, recall,
F1-score, and confusion matrix.

## 13. Model Evaluation

The trained machine learning models are evaluated using accuracy,
precision, recall, and F1-score.

These metrics help measure how well each model performs on the
unseen testing data.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

models = {
    "Logistic Regression": y_pred_logistic,
    "Decision Tree": y_pred_tree,
    "Random Forest": y_pred_forest
}

results = []

for model_name, predictions in models.items():
    results.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions, zero_division=0),
        "Recall": recall_score(y_test, predictions, zero_division=0),
        "F1 Score": f1_score(y_test, predictions, zero_division=0)
    })

results_df = pd.DataFrame(results)

results_df
```

```python
results_df.set_index("Model")[["Accuracy", "Precision", "Recall", "F1 Score"]].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Machine Learning Model Performance Comparison")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.show()
```

### Evaluation Summary

The performance of Logistic Regression, Decision Tree, and Random
Forest is compared using multiple evaluation metrics.

The evaluation results provide a quantitative comparison of the
classification models on the testing dataset.

## 14. Confusion Matrix

A confusion matrix is used to evaluate the classification performance
of each machine learning model.

It shows the number of correct and incorrect predictions for the two
classes: no heart disease and heart disease.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

model_predictions = {
    "Logistic Regression": y_pred_logistic,
    "Decision Tree": y_pred_tree,
    "Random Forest": y_pred_forest
}

for model_name, predictions in model_predictions.items():
    cm = confusion_matrix(y_test, predictions)

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["No Disease", "Disease"]
    )

    disp.plot()
    plt.title(f"Confusion Matrix - {model_name}")
    plt.show()
```

### Confusion Matrix Interpretation

The confusion matrix contains four possible outcomes:

- True Positive (TP): Correctly predicted heart disease.
- True Negative (TN): Correctly predicted no heart disease.
- False Positive (FP): Predicted heart disease when it was absent.
- False Negative (FN): Predicted no heart disease when it was present.

These values help us understand the types of errors made by each model.

## 15. ROC Curve and AUC Score

The Receiver Operating Characteristic (ROC) curve is used to evaluate
the classification performance of the machine learning models across
different classification thresholds.

The Area Under the Curve (AUC) summarizes the ROC curve into a single
value.

```python
from sklearn.metrics import roc_curve, roc_auc_score

# Get probability predictions
y_prob_logistic = logistic_model.predict_proba(X_test)[:, 1]
y_prob_tree = decision_tree_model.predict_proba(X_test)[:, 1]
y_prob_forest = random_forest_model.predict_proba(X_test)[:, 1]

# Calculate ROC curves
fpr_logistic, tpr_logistic, _ = roc_curve(y_test, y_prob_logistic)
fpr_tree, tpr_tree, _ = roc_curve(y_test, y_prob_tree)
fpr_forest, tpr_forest, _ = roc_curve(y_test, y_prob_forest)

# Calculate AUC scores
auc_logistic = roc_auc_score(y_test, y_prob_logistic)
auc_tree = roc_auc_score(y_test, y_prob_tree)
auc_forest = roc_auc_score(y_test, y_prob_forest)

print("Logistic Regression AUC:", round(auc_logistic, 3))
print("Decision Tree AUC:", round(auc_tree, 3))
print("Random Forest AUC:", round(auc_forest, 3))
```

```python
plt.figure(figsize=(9, 6))

plt.plot(
    fpr_logistic,
    tpr_logistic,
    label=f"Logistic Regression (AUC = {auc_logistic:.3f})"
)

plt.plot(
    fpr_tree,
    tpr_tree,
    label=f"Decision Tree (AUC = {auc_tree:.3f})"
)

plt.plot(
    fpr_forest,
    tpr_forest,
    label=f"Random Forest (AUC = {auc_forest:.3f})"
)

plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()
plt.show()
```

### ROC-AUC Summary

The ROC curves and AUC scores provide an additional measure of the
classification performance of the three machine learning models.

The AUC values are used together with accuracy, precision, recall,
F1-score, and the confusion matrices when interpreting model
performance.

## 17. New Patient Heart Disease Prediction

The trained machine learning model can be used to generate a prediction
for a new patient based on the patient's available health-related
attributes.

This demonstration is intended only for academic purposes and should
not be used as a medical diagnosis.

```python
def predict_heart_disease(patient_data):
    """
    Predict heart disease outcome for a new patient.
    """

    patient_df = pd.DataFrame([patient_data])

    # Apply the same preprocessing used during training
    patient_processed = preprocessor.transform(patient_df)

    # Select the trained model
    if selected_model_name == "Logistic Regression":
        model = logistic_model
    elif selected_model_name == "Decision Tree":
        model = decision_tree_model
    else:
        model = random_forest_model

    prediction = model.predict(patient_processed)[0]

    if prediction == 1:
        return "Heart disease predicted"
    else:
        return "No heart disease predicted"
```

### Prediction Result

The trained machine learning model processes the sample patient's
features and generates a predicted heart disease class.

The prediction is intended for demonstrating the machine learning
workflow and is not a substitute for professional medical evaluation.
